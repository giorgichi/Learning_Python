### If Statements


```python
n = input("Choose an intereger between 2 and 22")
n = int(n)
if n < 5:
    print("The integer that you choose is less than 5")
if n > 10:
    print("The integer is greater than 10")
else:
    print("The integer is less than 10")
```

## If Else statement

```python
def minimum(x, y):
    if x < y:
        return x
    else:
        return y
```

```python
result_min1 = minimum(-4, 3.6)
print(result_min1)
```

```python
result_min2 = minimum(33, 11)
print(result_min2)
```
