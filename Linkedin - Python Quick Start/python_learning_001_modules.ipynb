### Types of Sequence

There are multiple types of sequences in python:
- **List** = a sequence with a specific order
    - use square brackets `[ ]` to create and separate items by commas
    - can contain elements of different type
    - you _can_ change a value of element
- **Tuples** = a sequence with a specific order
    - use parentheses `( )` to create and separate items by commas
    - items _cannot be_ changed
      

#### Lists


```python
listA = [1, 4, 5, 64]
type(listA)
```

```python
print(listA)
print(listA[0])
```

```python
listA[2] = 20
print(listA)
```

```python
listB = [True, "star", 21, 3.14]
print(listB)
```
