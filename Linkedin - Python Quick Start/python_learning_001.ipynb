### Function that recives an input and returns an output


```python
def square(x):
    print("Gio can you square x?")
    return x * x
```

```python
result = square(5)
```

```python
print(result)
```

```python
anotherOne = square(result)
print(anotherOne)
```

### Function with two inputs


```python
def sumSquare(x, y):
    sq1 = x * x
    sq2 = y * y
    return sq1 + sq2
```

```python
sumSquare(2, 3)
```
