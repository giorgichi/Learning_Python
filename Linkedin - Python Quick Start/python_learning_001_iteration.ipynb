## Iteration in Python


```python
# Count Vowels in a word
word = "serendipitious"
vowels = ["a", "e", "o", "i", "u"]
vowelCount = 0

for char in word:
    if char in vowels:
        vowelCount += 1

print(vowelCount)
```

## Range in For Loop

```python
for i in range(4):
    print(i)
```

```python
for i in range (1,8,2):
    print(i)
```

```python
for i in vowels:
    print(i)
```

## While Loop


```python
len(vowels)
```

```python
i = 0
while i < len(vowels):
    print(vowels[i])
    i += 1
```
