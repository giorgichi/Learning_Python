```python
import numpy as np
import matplotlib.pyplot as pp
```

```python
%matplotlib inline
```

```python
x = np.linspace(0, 10, 60)
x
```

```python
sinx = np.sin(x)
sinx
```

```python
pp.plot(x, sinx)
```

```python
cosx = np.cos(x)
```

```python
pp.plot(x, sinx)
pp.plot(x, cosx)
```

```python
pp.plot(x, sinx, 'o')
pp.plot(x, cosx, 'x')
```

```python
y = sinx * cosx
z = cosx**2 - sinx**2

pp.plot(x, y, 'o')
pp.plot(x, z, 'red')
pp.plot(x, sinx, 'black')

pp.legend(['sin(x) cos(x)', 'cos(x)^2 - sin(x)^2'])
```
