**DEDUPLICATE**

One way to remove duplicates is to convert _list_ into **Python Set** and then convert back to list

```python
wordsunique = list(set(wordclean))
wordsunique[:10]
```

```python
wordsunique.sort()
```

```python
wordsunique[:10]
```

```python
wordclean = [word.strip().lower() for word in open('words', 'r')] 
```

```python
wordclean = sorted(list(set([word.strip().lower() for word in open('words', 'r')])))
```
