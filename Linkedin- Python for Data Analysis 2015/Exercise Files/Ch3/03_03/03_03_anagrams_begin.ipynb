```python
def signature(word):
    return ''.join(sorted(word))
```

```python
signature('lives')
```

```python
'/'.join("gio")
```

```python
def anagram(myword):
    return [word for word in wordclean if signature(word) == signature(myword)]
```

```python
anagram('dictionary')
```

```python
%timeit anagram('dictionary')
```

```python
words_bysig = {}

for word in wordclean:
    words_bysig[signature(word)].append(word)
```

```python
import collections
```

```python
words_bysig = collections.defaultdict(list)

for word in wordclean:
    words_bysig[signature(word)].append(word)
```

### Back to Finding anagrams using new dicrionary


```python
def anagram_fast(myword):
    return words_bysig[signature(myword)]
```

```python
anagram_fast("dance")
```

#### Get all the anagrams in the dictionary 
except those cases when the word is an anagram of itself

```python
anagrams_all = {word: anagram_fast(word) for word in wordclean
                if len(anagram_fast(word)) > 1}
```

```python
len(anagrams_all)
```

```python
anagrams_gio = {word: anagram_fast(word) for word in wordclean
                if len(word) == 22 and len(anagram_fast(word)) > 1}
len(anagrams_gio)
```
