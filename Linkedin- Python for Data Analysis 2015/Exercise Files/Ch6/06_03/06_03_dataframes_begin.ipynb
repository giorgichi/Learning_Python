```python
presidents = pd.DataFrame([
    {'name':'Barack Obama', 'inaguration':2009, 'birthyear':1961},    
    {'name':'George W. Bush', 'inaguration':2001, 'birthyear':1946},
    {'name':'Bill Clinton', 'birthyear':1946, 'inaguration':1993},
    {'name':'George H. W. Bush', 'inaguration':1989, 'birthyear':1924}
])
presidents
```

```python
presidents_index = presidents.set_index('name')
presidents_index
```

```python
presidents_fathers = pd.DataFrame([
    {'son':'Barack Obama', 'father':'Barack Obama Sr.'},
    {'son':'George W. Bush', 'father':'George H. W. Bush'},
    #{'son':'Bill Clinton', 'father':''},
    {'son':'George H. W. Bush', 'father':'Presccot Bush'}
])
presidents_fathers
```

```python
pd.merge(presidents, presidents_fathers, left_on='name', right_on='son')
```

```python
pd.merge(presidents, presidents_fathers, left_on='name', right_on='son').drop('son', axis=1)
```

```python
pd.merge(presidents, presidents_fathers, left_on='name', right_on='son', how='outer').drop('son', axis=1)
```
