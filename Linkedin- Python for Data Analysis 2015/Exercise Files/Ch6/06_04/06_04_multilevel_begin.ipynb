```python
import pandas as pd
import seaborn
```

```python
flights = seaborn.load_dataset('flights')
flights.head()
```

```python
fl_indexed = flights.set_index(['year', 'month'])
fl_indexed.head()
```

```python
fl_indexed.loc[1950]
```

```python
fl_unstacked = fl_indexed.unstack()
```

```python
fl_unstacked
```

```python
fl_unstacked.sum(axis=0)
```

```python
fl_unstacked.sum(axis=1)
```

```python
fl_unstacked['passengers','total'] = fl_unstacked.sum(axis=1)
```

```python
fl_unstacked
```

```python
fl_restacked = fl_unstacked.stack()
```
