```python
def topten(sex,year):
    simple = allyears_indexed.loc[sex,:,year].sort_values('number',ascending=False).reset_index()
    simple = simple.drop(['number'],axis=1).head(10)
    
    simple.columns = [year]
    simple.index = simple.index + 1
    
    return simple
```

```python
topten('M',2009)
```

```python
def toptens(sex,year0,year1):
    years = [topten(sex,year) for year in range(year0,year1+1)]
    
    return years[0].join(years[1:])
```

```python
toptens('M',2000,2010)
```

```python
# trying to compute how spiky the data (popularity of name) is = name fads
# by squaring the counts (multiply number by itself) will increase the contrast

# First - compute total number of babes with a given names over all years

allyears.groupby(['sex', 'name']).sum().head()
```

```python
totals = allyears.groupby(['sex', 'name'])['number'].sum()
totals
```

```python
def sumsq(x):
    return sum(x**2)
```

```python
spikyness = allyears.groupby(['sex','name'])['number'].agg(sumsq) / totals**2
```

```python
spikyness.head()
```

```python
spiky_common = spikyness[totals > 5000].copy()
spiky_common.sort_values(ascending=False)
spiky_common.head(10)
```

```python
spiky_common.sort_values()
```

```python
pp.figure(figsize=(12,3))

plotname('F', 'Louisa')
plotname('M', 'Shaquille')
```

```python
fads = spiky_common.sort_values().tail(10).index.values
fads
```

```python
pp.figure(figsize=(12,3.4))

for sex, name in fads:
    plotname(sex, name)

pp.legend([name for sex, name in fads], loc='upper left')
```

```python
total_recent = allyears[allyears['year']>2005].groupby(['sex','name'])['number'].sum()
```

```python
# Re-create MultiIndexes to clear stale C-engine metadata
spikyness.index = pd.MultiIndex.from_tuples(spikyness.index)
totals.index = pd.MultiIndex.from_tuples(totals.index)
```

```python
spiky_common = spikyness[(totals > 5000) & (total_recent < 1000)].copy()
```

```python
fads = spiky_common.sort_values(ascending=False).head(10).index.values
```

```python
fads
```

```python
pp.figure(figsize=(12,2.5))

for sex,name in fads:
    plotname(sex,name)

pp.legend([name for sex,name in fads],loc='upper left')
```
