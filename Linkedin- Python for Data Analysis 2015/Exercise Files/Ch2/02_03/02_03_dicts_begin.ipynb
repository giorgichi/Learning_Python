## Dictionaries

Alphabetic order is not garaneeted for keys and values of dictionaries, index is not garnteed.

1. Flexible and efficient containers to associate labels with heterogenous data
2. Use where data items can be given labels
3. Most appropriate for collecting data iteamd of different kinds

```python
capitals = {'United States':'Washington, DC', 'France':'Paris', 'Italy':'Rome'}
```

```python
capitals
```

```python
capitals['Italy']
```

```python
capitals['Spain'] = "Madrid"
```

```python
capitals["Spain"]
```

```python
capitals
```

```python
'Italy' in capitals
```

```python
'Germany' in capitals
```

```python
morecapitals = {"Germany":"Berlin", "Spain":"Madrid"}
```

```python
capitals.update(morecapitals)
```

```python
capitals
```

```python
del capitals["United States"]
```

```python
capitals
```

```python
for key in capitals:
    print(capitals[key], "is in", key)
```

```python
for value in capitals.values():
    print(value)
```

```python
for key, value in capitals.items():
    print(value, 'is in', key)
```
