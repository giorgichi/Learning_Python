## Object Oriented Programming (OOP)

Convention is to Capitalize first latter of the name for class 

```python
class Dog():
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def bark(self):
        print("Woof, woof!")
```

Creating an instance of a class is called **"instantiation"** 

```python
fido = Dog("Fido", 2)
```

```python
fido.bark()
```

```python
print(fido.age)
```

```python
print(fido.name)
```

```python
fido.age += 1
```

```python
fido.age
```
